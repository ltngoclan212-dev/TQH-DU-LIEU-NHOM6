# 3. Trực quan hóa dữ liệu bằng Plotly

## 3.1 Histogram – Phân phối dữ liệu

**Mục tiêu:** Quan sát cách các giá trị doanh thu du lịch lữ hành phân bố giữa các địa phương trong toàn bộ giai đoạn.

Histogram phù hợp vì giúp nhìn được:
- Tập trung dữ liệu ở khoảng nào.
- Phân phối có lệch hay không.
- Có xuất hiện các giá trị rất lớn so với phần còn lại hay không.


```python
# 3.1. Histogram: Phân phối doanh thu du lịch lữ hành

hist_df = df_dia_phuong.dropna(subset=["Doanh thu du lịch lữ hành (Tỷ đồng)"]).copy()

fig_hist = px.histogram(
    hist_df,
    x="Doanh thu du lịch lữ hành (Tỷ đồng)",
    nbins=30,
    marginal="box",
    title="Phân phối doanh thu du lịch lữ hành theo địa phương",
    labels={
        "Doanh thu du lịch lữ hành (Tỷ đồng)": "Doanh thu (tỷ đồng)"
    },
    hover_data=["Địa phương", "Năm"]
)

fig_hist.update_layout(
    bargap=0.05,
    template="plotly_white"
)

fig_hist.show()
```

### Nhận xét Histogram

Histogram cho biết phần lớn các quan sát doanh thu tập trung ở vùng giá trị thấp hơn, trong khi một số địa phương có doanh thu rất cao tạo ra phần đuôi bên phải của phân phối. Điều này cho thấy doanh thu du lịch giữa các địa phương có sự chênh lệch đáng kể.

**Insight:** Phân phối lệch phải là dấu hiệu cho thấy một nhóm nhỏ địa phương có quy mô doanh thu nổi bật so với phần lớn các địa phương còn lại.


## 3.2 Box Plot – Phân bố và giá trị ngoại lệ

**Mục tiêu:** Đánh giá trung vị, khoảng biến thiên và các giá trị ngoại lệ của doanh thu du lịch lữ hành.

Box Plot đặc biệt hữu ích khi muốn phát hiện các địa phương có doanh thu cao bất thường so với mặt bằng chung.


```python
# 3.2. Box Plot: Phân bố doanh thu

box_df = df_dia_phuong.dropna(subset=["Doanh thu du lịch lữ hành (Tỷ đồng)"]).copy()

fig_box = px.box(
    box_df,
    y="Doanh thu du lịch lữ hành (Tỷ đồng)",
    points="outliers",
    title="Box Plot doanh thu du lịch lữ hành",
    labels={
        "Doanh thu du lịch lữ hành (Tỷ đồng)": "Doanh thu (tỷ đồng)"
    },
    hover_data=["Địa phương", "Năm"]
)

fig_box.update_layout(
    template="plotly_white"
)

fig_box.show()
```

### Nhận xét Box Plot

Box Plot cho thấy phần lớn địa phương có doanh thu nằm trong một khoảng tương đối thấp, trong khi một số giá trị nằm rất xa phần còn lại. Các điểm ngoại lệ này có thể tương ứng với những trung tâm du lịch lớn.

**Lưu ý:** Giá trị ngoại lệ trong biểu đồ không đồng nghĩa với dữ liệu sai. Đây có thể là những quan sát thực tế có quy mô lớn và cần được giữ lại nếu dữ liệu nguồn đáng tin cậy.


## 3.3 Stacked Bar – Cơ cấu khách trong nước và quốc tế

**Mục tiêu:** So sánh quy mô và cơ cấu giữa khách trong nước và khách quốc tế theo từng năm.

Stacked Bar phù hợp vì vừa thể hiện **tổng quy mô khách**, vừa cho thấy **đóng góp của từng nhóm khách** vào tổng số.


```python
# 3.3. Stacked Bar: Khách trong nước và khách quốc tế

stack_df = df_nam[["Năm", "Khách trong nước - cơ sở lưu trú (Nghìn lượt)", "Khách quốc tế - cơ sở lưu trú (Nghìn lượt)"]].copy()

stack_long = stack_df.melt(
    id_vars="Năm",
    value_vars=["Khách trong nước - cơ sở lưu trú (Nghìn lượt)", "Khách quốc tế - cơ sở lưu trú (Nghìn lượt)"],
    var_name="Nhóm khách",
    value_name="Lượt khách"
)

fig_stack = px.bar(
    stack_long,
    x="Năm",
    y="Lượt khách",
    color="Nhóm khách",
    barmode="stack",
    title="Cơ cấu khách trong nước và khách quốc tế theo năm",
    labels={
        "Lượt khách": "Lượt khách (nghìn lượt)",
        "Năm": "Năm",
        "Nhóm khách": "Nhóm khách"
    },
    text_auto=".2s"
)

fig_stack.update_layout(
    template="plotly_white",
    hovermode="x unified"
)

fig_stack.show()
```

### Nhận xét Stacked Bar

Biểu đồ cho thấy khách trong nước chiếm phần lớn tổng lượng khách lưu trú trong hầu hết các năm. Giai đoạn 2020–2021 có sự suy giảm rõ rệt ở cả hai nhóm, trong đó khách quốc tế biến động mạnh hơn. Từ năm 2022, tổng lượng khách phục hồi và tăng nhanh.

**Insight:** Cơ cấu khách cho thấy thị trường nội địa đóng vai trò quan trọng trong quy mô du lịch, đồng thời thị trường quốc tế là nhóm có mức biến động lớn và thể hiện rõ quá trình phục hồi sau giai đoạn suy giảm.


# 4. Tổng hợp phát hiện chính

Histogram cho thấy doanh thu giữa các địa phương phân bố không đồng đều và có xu hướng lệch phải, với một số địa phương có doanh thu rất cao.

### Phát hiện 2 – Giá trị ngoại lệ

### Phát hiện 3 – Cơ cấu khách
Stacked Bar cho thấy khách trong nước chiếm tỷ trọng lớn trong tổng lượng khách, trong khi khách quốc tế biến động mạnh hơn theo thời gian.

### Phát hiện 4 – Giai đoạn biến động
Cả hai nhóm khách đều giảm mạnh trong giai đoạn 2020–2021 và phục hồi rõ rệt từ năm 2022.
