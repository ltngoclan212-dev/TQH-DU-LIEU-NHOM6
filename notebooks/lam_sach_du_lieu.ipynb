```python
from pathlib import Path
import pandas as pd
import numpy as np

ROOT = Path.cwd()

# Nếu notebook đang chạy từ thư mục notebooks/
if not (ROOT / "data" / "raw").exists():
    ROOT = ROOT.parent

RAW = ROOT / "data" / "raw"
PROCESSED = ROOT / "data" / "processed"

PROCESSED.mkdir(parents=True, exist_ok=True)

print("ROOT:", ROOT)
print("RAW :", RAW)
print("PROCESSED:", PROCESSED)
```

```python
df_nam_raw = pd.read_csv(
    RAW / "du_lieu_du_lich_nam.csv"
)

df_dp_raw = pd.read_csv(
    RAW / "du_lieu_du_lich_dia_phuong.csv"
)

print("Dataset theo năm:", df_nam_raw.shape)
print("Dataset địa phương:", df_dp_raw.shape)

display(df_nam_raw.head())
display(df_dp_raw.head())
```

```python
print("===== DATASET THEO NĂM =====")
print(df_nam_raw.info())

print("\nMissing:")
print(df_nam_raw.isnull().sum())

print("\nDuplicate:", df_nam_raw.duplicated().sum())


print("\n\n===== DATASET ĐỊA PHƯƠNG =====")
print(df_dp_raw.info())

print("\nMissing:")
print(df_dp_raw.isnull().sum())

print("\nDuplicate:", df_dp_raw.duplicated().sum())
```

```python
def dem_ky_hieu_hai_cham(df):
    return (
        df.astype(str)
          .apply(lambda col: col.str.strip().eq(".."))
          .sum()
    )

print("Số ký hiệu '..' trong dataset theo năm:")
print(dem_ky_hieu_hai_cham(df_nam_raw))

print("\nSố ký hiệu '..' trong dataset địa phương:")
print(dem_ky_hieu_hai_cham(df_dp_raw))
```

```python
df_nam = df_nam_raw.copy()
df_dp = df_dp_raw.copy()

# Chuẩn hóa tên cột
df_nam.columns = df_nam.columns.str.strip()
df_dp.columns = df_dp.columns.str.strip()

# Chuẩn hóa khoảng trắng trong các cột dạng text
for col in df_nam.select_dtypes(include="object").columns:
    df_nam[col] = (
        df_nam[col]
        .astype(str)
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

for col in df_dp.select_dtypes(include="object").columns:
    df_dp[col] = (
        df_dp[col]
        .astype(str)
        .str.strip()
        .str.replace(r"\s+", " ", regex=True)
    )

# Chuyển ký hiệu nguồn ".." thành dữ liệu thiếu
df_nam = df_nam.replace("..", np.nan)
df_dp = df_dp.replace("..", np.nan)

print("Đã chuẩn hóa khoảng trắng và xử lý ký hiệu '..'.")
```

```python
# Lưu trạng thái của năm trước khi chuẩn hóa
df_nam["Trạng thái"] = df_nam["Năm"].astype(str).apply(
    lambda x: "Sơ bộ" if "Sơ bộ" in x else ""
)

# Tách năm 4 chữ số: "Sơ bộ 2024" -> 2024
df_nam["Năm"] = (
    df_nam["Năm"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
    .astype("int64")
)

# Các cột cần chuyển sang numeric
# Loại trừ Năm và Trạng thái
cot_so_nam = [
    col for col in df_nam.columns
    if col not in ["Năm", "Trạng thái"]
]

for col in cot_so_nam:
    df_nam[col] = pd.to_numeric(
        df_nam[col],
        errors="coerce"
    )

print(df_nam.dtypes)
display(df_nam[["Năm", "Trạng thái"]])
```

```python
# Năm phải là số nguyên
df_dp["Năm"] = pd.to_numeric(
    df_dp["Năm"],
    errors="raise"
).astype("int64")

# Doanh thu phải là numeric
df_dp["Doanh thu du lịch lữ hành (Tỷ đồng)"] = pd.to_numeric(
    df_dp["Doanh thu du lịch lữ hành (Tỷ đồng)"],
    errors="coerce"
)

# Chuẩn hóa các cột text quan trọng
for col in ["Địa phương", "Cấp dữ liệu", "Trạng thái"]:
    if col in df_dp.columns:
        df_dp[col] = (
            df_dp[col]
            .astype("string")
            .str.strip()
            .str.replace(r"\s+", " ", regex=True)
        )

print(df_dp.dtypes)
```

```python
print("Duplicate dataset năm trước:", df_nam.duplicated().sum())
print("Duplicate dataset địa phương trước:", df_dp.duplicated().sum())

df_nam = df_nam.drop_duplicates().reset_index(drop=True)
df_dp = df_dp.drop_duplicates().reset_index(drop=True)

print("\nSau xử lý:")
print("Dataset năm:", df_nam.duplicated().sum())
print("Dataset địa phương:", df_dp.duplicated().sum())
```

```python
print("Các cấp dữ liệu:")
print(df_dp["Cấp dữ liệu"].value_counts(dropna=False))

print("\nCác trạng thái:")
print(df_dp["Trạng thái"].value_counts(dropna=False))

print("\nCác năm:")
print(sorted(df_dp["Năm"].unique()))

print("\nSố địa phương khác nhau:")
print(df_dp["Địa phương"].nunique())
```

```python
print("===== DATASET NĂM =====")
print("Shape:", df_nam.shape)
print("Duplicate:", df_nam.duplicated().sum())
print("\nMissing:")
print(df_nam.isnull().sum())


print("\n===== DATASET ĐỊA PHƯƠNG =====")
print("Shape:", df_dp.shape)
print("Duplicate:", df_dp.duplicated().sum())
print("\nMissing:")
print(df_dp.isnull().sum())
```

```python
file_nam_clean = PROCESSED / "du_lieu_du_lich_nam_clean.csv"
file_dp_clean = PROCESSED / "du_lieu_du_lich_dia_phuong_clean.csv"

df_nam.to_csv(
    file_nam_clean,
    index=False,
    encoding="utf-8-sig"
)

df_dp.to_csv(
    file_dp_clean,
    index=False,
    encoding="utf-8-sig"
)

print("Đã tạo:")
print(file_nam_clean)
print(file_dp_clean)
```
