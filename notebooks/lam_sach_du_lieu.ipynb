## 4. Kiểm tra ký hiệu dữ liệu thiếu

Ngoài giá trị `NaN`, dữ liệu nguồn có thể sử dụng ký hiệu `..` để biểu diễn số liệu chưa có hoặc chưa được công bố.

Pandas không tự động xem `..` là dữ liệu thiếu, vì vậy cần kiểm tra riêng trước khi xử lý.

Các giá trị này không được tự động thay bằng `0`, vì:

- `0` có nghĩa là giá trị thực sự bằng không;
- dữ liệu thiếu có nghĩa là chưa có hoặc chưa được công bố.

```python
def dem_ky_hieu_hai_cham(df):
    ket_qua = {}

    for col in df.columns:
        count = (
            df[col]
            .astype("string")
            .str.strip()
            .eq("..")
            .sum()
        )

        if count > 0:
            ket_qua[col] = int(count)

    return pd.Series(ket_qua, dtype="int64")


print("===== KÝ HIỆU '..' - DATASET THEO NĂM =====")
display(dem_ky_hieu_hai_cham(df_nam_raw))

print("\n===== KÝ HIỆU '..' - DATASET ĐỊA PHƯƠNG =====")
display(dem_ky_hieu_hai_cham(df_dp_raw))
```

## 5. Tạo bản sao và chuẩn hóa dữ liệu

Hai dataset gốc được sao chép trước khi làm sạch để không làm thay đổi dữ liệu trong `data/raw/`.

Các bước xử lý:

1. Loại bỏ khoảng trắng dư trong tên cột.
2. Loại bỏ BOM nếu có.
3. Chuẩn hóa khoảng trắng trong dữ liệu dạng chuỗi.
4. Chuyển ký hiệu `..` thành dữ liệu thiếu.

```python
df_nam = df_nam_raw.copy()
df_dp = df_dp_raw.copy()


def chuan_hoa_co_ban(df):
    df = df.copy()

    # Chuẩn hóa tên cột
    df.columns = (
        df.columns
        .str.replace("\ufeff", "", regex=False)
        .str.strip()
    )

    # Chuẩn hóa dữ liệu dạng text nhưng vẫn giữ missing
    text_cols = df.select_dtypes(
        include=["object", "string"]
    ).columns

    for col in text_cols:
        df[col] = (
            df[col]
            .astype("string")
            .str.strip()
            .str.replace(r"\s+", " ", regex=True)
        )

    # ".." -> missing
    df = df.replace("..", pd.NA)

    return df


df_nam = chuan_hoa_co_ban(df_nam)
df_dp = chuan_hoa_co_ban(df_dp)

print("Đã chuẩn hóa dữ liệu cơ bản.")
```

## 6. Chuẩn hóa dữ liệu du lịch theo năm

Trong dữ liệu gốc, năm 2024 được biểu diễn dưới dạng `Sơ bộ 2024`.

Để thuận tiện cho phân tích:

- cột `Năm` được chuẩn hóa thành số nguyên;
- tạo thêm cột `Trạng thái`;
- 2015–2023 được ghi là `Chính thức`;
- 2024 được ghi là `Sơ bộ`.

Các chỉ tiêu doanh thu và lượng khách được chuyển sang kiểu số.

```python
# Giữ lại nội dung gốc của cột Năm
nam_text = df_nam["Năm"].astype("string")

# Tạo cột Trạng thái
df_nam["Trạng thái"] = np.where(
    nam_text.str.contains("Sơ bộ", case=False, na=False),
    "Sơ bộ",
    "Chính thức"
)

# "Sơ bộ 2024" -> 2024
df_nam["Năm"] = pd.to_numeric(
    nam_text.str.extract(r"(\d{4})")[0],
    errors="raise"
).astype("int64")

# Các cột còn lại phải là numeric
cot_so_nam = [
    col
    for col in df_nam.columns
    if col not in ["Năm", "Trạng thái"]
]

for col in cot_so_nam:
    df_nam[col] = pd.to_numeric(
        df_nam[col],
        errors="coerce"
    )

print("Kiểu dữ liệu sau chuẩn hóa:")
print(df_nam.dtypes)

print("\nNăm và trạng thái:")
display(df_nam[["Năm", "Trạng thái"]])
```

## 7. Chuẩn hóa dữ liệu du lịch theo địa phương

Bộ dữ liệu địa phương được chuẩn hóa:

- `Năm` thành kiểu số nguyên;
- doanh thu thành dữ liệu số;
- `Địa phương`, `Cấp dữ liệu` và `Trạng thái` thành dữ liệu chuỗi;
- năm 2024 được xác định là số liệu sơ bộ;
- các năm trước được xác định là số liệu chính thức nếu trạng thái chưa được ghi rõ.

```python
# Chuẩn hóa cột Năm
df_dp["Năm"] = pd.to_numeric(
    df_dp["Năm"]
    .astype("string")
    .str.extract(r"(\d{4})")[0],
    errors="raise"
).astype("int64")

# Chuẩn hóa doanh thu
cot_doanh_thu = "Doanh thu du lịch lữ hành (Tỷ đồng)"

df_dp[cot_doanh_thu] = pd.to_numeric(
    df_dp[cot_doanh_thu],
    errors="coerce"
)

# Chuẩn hóa các cột text
for col in ["Địa phương", "Cấp dữ liệu", "Trạng thái"]:
    if col in df_dp.columns:
        df_dp[col] = (
            df_dp[col]
            .astype("string")
            .str.strip()
            .str.replace(r"\s+", " ", regex=True)
        )

# Nếu chưa có cột Trạng thái thì tạo mới
if "Trạng thái" not in df_dp.columns:
    df_dp["Trạng thái"] = pd.NA

# Chuỗi rỗng -> missing
df_dp["Trạng thái"] = df_dp["Trạng thái"].replace("", pd.NA)

# Điền trạng thái còn thiếu dựa trên năm
df_dp["Trạng thái"] = df_dp["Trạng thái"].fillna(
    pd.Series(
        np.where(
            df_dp["Năm"].eq(2024),
            "Sơ bộ",
            "Chính thức"
        ),
        index=df_dp.index
    )
)

print(df_dp.dtypes)

display(df_dp.head())
```

## 8. Kiểm tra bản ghi trùng lặp

Các bản ghi giống hoàn toàn nhau có thể làm sai lệch kết quả thống kê và trực quan hóa.

Nếu phát hiện bản ghi trùng hoàn toàn, các bản ghi lặp được loại bỏ.

Không loại bỏ các dòng chỉ vì chúng có cùng năm hoặc cùng địa phương, vì đó có thể là những quan sát hợp lệ.

```python
duplicate_nam = df_nam.duplicated().sum()
duplicate_dp = df_dp.duplicated().sum()

print("Duplicate dataset theo năm:", duplicate_nam)
print("Duplicate dataset địa phương:", duplicate_dp)

df_nam = (
    df_nam
    .drop_duplicates()
    .reset_index(drop=True)
)

df_dp = (
    df_dp
    .drop_duplicates()
    .reset_index(drop=True)
)

print("\nSau xử lý:")
print("Dataset theo năm:", df_nam.duplicated().sum())
print("Dataset địa phương:", df_dp.duplicated().sum())
```
